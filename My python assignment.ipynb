````python
import pandas as pd

# Creating the DataFrame
data = {
    "Age Group": ["15-19", "20-24", "25-29", "30-34"],
    "Women Hypertension": [3, 6, 10, 11],
    "Men Hypertension": [1, 2, 2, 5],
    "Women Diabetes": [1, 1, 1, 1],
    "Men Diabetes": [0, 1, 1, 1],
}

df = pd.DataFrame(data)

# Grouping data by 'Age Group' and calculating totals and averages
grouped_totals = df.groupby("Age Group").sum(numeric_only=True)
grouped_averages = df.groupby("Age Group").mean(numeric_only=True)

# Presenting results
print("Total sum")
print(grouped_totals)

print("\n Total avarage")
print(grouped_averages)
````

````python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

# Creating the DataFrame
data = {
    "Age Group": ["15-19", "20-24", "25-29", "30-34"],
    "Women Hypertension": [3, 6, 10, 11],
    "Men Hypertension": [1, 2, 2, 5],
}

df = pd.DataFrame(data)

# Bar Chart: Total Hypertension Cases by Gender
totals = {
    "Gender": ["Women", "Men"],
    "Total Cases": [df["Women Hypertension"].sum(), df["Men Hypertension"].sum()],
}
totals_df = pd.DataFrame(totals)

plt.figure(figsize=(8, 6))
sns.barplot(x="Gender", y="Total Cases", data=totals_df, palette="Blues_d")
plt.title("Total Hypertension Cases by Gender", fontsize=16)
plt.xlabel("Gender", fontsize=12)
plt.ylabel("Total Cases", fontsize=12)
plt.tight_layout()
plt.show()

# Line Graph: Hypertension Trends Across Age Groups
plt.figure(figsize=(10, 6))
sns.lineplot(x="Age Group", y="Women Hypertension", data=df, marker="o", label="Women", linewidth=2, color="blue")
sns.lineplot(x="Age Group", y="Men Hypertension", data=df, marker="o", label="Men", linewidth=2, color="green")

plt.title("Hypertension Trends Across Age Groups", fontsize=16)
plt.xlabel("Age Group", fontsize=12)
plt.ylabel("Number of Cases", fontsize=12)
plt.legend(title="Gender", fontsize=10)
plt.tight_layout()
plt.show()
````

````python
# Hypertension Data Visualization

This project analyzes and visualizes hypertension cases for women and men across different age groups using Python. It provides insights into total cases and trends using bar charts and line graphs.

## Overview

The project performs the following tasks:
1. **Data Analysis**:
   - Calculates total hypertension cases for women and men.
2. **Data Visualization**:
   - A **bar chart** showing the total hypertension cases by gender.
   - A **line graph** illustrating trends in hypertension cases across age groups for both genders.

## Dataset

The dataset contains information on hypertension cases broken down by:
- **Age Group**: Categories such as `15-19`, `20-24`, `25-29`, `30-34`.
- **Women Hypertension**: Number of cases in women.
- **Men Hypertension**: Number of cases in men.

### Example Data:

| Age Group | Women Hypertension | Men Hypertension |
|-----------|---------------------|------------------|
| 15-19     | 3                   | 1                |
| 20-24     | 6                   | 2                |
| 25-29     | 10                  | 2                |
| 30-34     | 11                  | 5                |

## Visualizations

1. **Bar Chart**: 
   - Displays the total number of hypertension cases by gender.
   - Highlights gender-based differences in hypertension cases.

2. **Line Graph**: 
   - Shows trends in hypertension cases across age groups for women and men.
   - Compares gender-specific trends over time.

### Example Charts:
- **Bar Chart**: Visualizes total hypertension cases for women and men.
- **Line Graph**: Trends of hypertension cases across age groups.

## Requirements

The following libraries are required to run the code:
- **Pandas**: For data manipulation.
- **Matplotlib**: For creating visualizations.
- **Seaborn**: For advanced visualizations.

Install dependencies using pip:

```bash
pip install pandas matplotlib seaborn

## File structure                                              
hypertension-visualization/
│
├── data/
│   ├── hypertension_data.csv   
│
├── scripts/
│   ├── visualize_hypertension.py    
│
├── LICENSE                       
├── README.md                      
├── requirements.txt              
└── .gitignore                   
````
